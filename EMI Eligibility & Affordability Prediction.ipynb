```python
# ============================================
# STEP 2: DATA CLEANING
# ============================================

import pandas as pd
import numpy as np

# Make a copy of original dataset
clean_df = df.copy()

print("Original Shape:", clean_df.shape)


# --------------------------------------------
# 1. Convert age to numeric
# --------------------------------------------

clean_df['age'] = pd.to_numeric(
    clean_df['age'],
    errors='coerce'
)


# --------------------------------------------
# 2. Convert monthly_salary to numeric
# --------------------------------------------

clean_df['monthly_salary'] = pd.to_numeric(
    clean_df['monthly_salary'],
    errors='coerce'
)


# --------------------------------------------
# 3. Convert bank_balance to numeric
# --------------------------------------------

clean_df['bank_balance'] = pd.to_numeric(
    clean_df['bank_balance'],
    errors='coerce'
)


# --------------------------------------------
# 4. Handle missing categorical value
# --------------------------------------------

clean_df['education'] = clean_df['education'].fillna(
    clean_df['education'].mode()[0]
)


# --------------------------------------------
# 5. Handle missing numerical values
# Using median
# --------------------------------------------

numeric_columns = [
    'age',
    'monthly_salary',
    'monthly_rent',
    'credit_score',
    'bank_balance',
    'emergency_fund'
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )


# --------------------------------------------
# 6. Check missing values again
# --------------------------------------------

print("\nMissing Values After Cleaning:")
print(clean_df.isnull().sum())


# --------------------------------------------
# 7. Check data types
# --------------------------------------------

print("\nData Types After Cleaning:")
print(clean_df.dtypes)


# --------------------------------------------
# 8. Check duplicate rows
# --------------------------------------------

print("\nDuplicate Rows:", clean_df.duplicated().sum())


# --------------------------------------------
# 9. Final shape
# --------------------------------------------

print("\nFinal Dataset Shape:", clean_df.shape)
```

```python
# ============================================
# STEP 4: FEATURE ENGINEERING
# ============================================

# Make another copy
model_df = clean_df.copy()

# --------------------------------------------
# 1. Total Monthly Expenses
# --------------------------------------------

model_df['total_monthly_expenses'] = (
    model_df['monthly_rent'] +
    model_df['school_fees'] +
    model_df['college_fees'] +
    model_df['travel_expenses'] +
    model_df['groceries_utilities'] +
    model_df['other_monthly_expenses']
)


# --------------------------------------------
# 2. Total Debt / EMI Burden
# --------------------------------------------

model_df['total_debt_burden'] = (
    model_df['current_emi_amount'] +
    model_df['max_monthly_emi']
)


# --------------------------------------------
# 3. Debt-to-Income Ratio
# --------------------------------------------

model_df['debt_to_income_ratio'] = (
    model_df['total_debt_burden'] /
    model_df['monthly_salary']
)


# --------------------------------------------
# 4. Expense-to-Income Ratio
# --------------------------------------------

model_df['expense_to_income_ratio'] = (
    model_df['total_monthly_expenses'] /
    model_df['monthly_salary']
)


# --------------------------------------------
# 5. Remaining Monthly Income
# --------------------------------------------

model_df['remaining_income'] = (
    model_df['monthly_salary'] -
    model_df['total_monthly_expenses'] -
    model_df['current_emi_amount']
)


# --------------------------------------------
# 6. Loan-to-Income Ratio
# --------------------------------------------

model_df['loan_to_income_ratio'] = (
    model_df['requested_amount'] /
    (model_df['monthly_salary'] * 12)
)


# --------------------------------------------
# 7. Financial Stability Score
# --------------------------------------------

model_df['financial_stability_score'] = (
    model_df['credit_score'] / 1200
)


# --------------------------------------------
# Check new features
# --------------------------------------------

new_features = [
    'total_monthly_expenses',
    'total_debt_burden',
    'debt_to_income_ratio',
    'expense_to_income_ratio',
    'remaining_income',
    'loan_to_income_ratio',
    'financial_stability_score'
]

print("New Features Created:")
print(new_features)

print("\nNew Feature Preview:")
display(model_df[new_features].head())

print("\nNew Dataset Shape:")
print(model_df.shape)
```

```python
# ============================================
# STEP 5: TRAIN / TEST SPLIT
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# --------------------------------------------
# CLASSIFICATION
# Target = emi_eligibility
# --------------------------------------------

classification_df = model_df.copy()

# Remove target leakage features
X_class = classification_df.drop(
    columns=[
        'emi_eligibility',
        'max_monthly_emi',
        'total_debt_burden',
        'debt_to_income_ratio'
    ]
)

y_class = classification_df['emi_eligibility']


# Identify categorical and numerical columns
categorical_class = X_class.select_dtypes(
    include=['object']
).columns.tolist()

numerical_class = X_class.select_dtypes(
    exclude=['object']
).columns.tolist()


print("Classification Features:", X_class.shape)
print("Categorical Features:", len(categorical_class))
print("Numerical Features:", len(numerical_class))


# Train-Test Split
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)


print("\nClassification Training Data:", X_train_class.shape)
print("Classification Testing Data:", X_test_class.shape)
```

```python
# ============================================
# REGRESSION
# Target = max_monthly_emi
# ============================================

regression_df = model_df.copy()

# Remove regression target and target-derived features
X_reg = regression_df.drop(
    columns=[
        'max_monthly_emi',
        'emi_eligibility',
        'total_debt_burden',
        'debt_to_income_ratio'
    ]
)

y_reg = regression_df['max_monthly_emi']


# Identify categorical and numerical columns
categorical_reg = X_reg.select_dtypes(
    include=['object']
).columns.tolist()

numerical_reg = X_reg.select_dtypes(
    exclude=['object']
).columns.tolist()


print("Regression Features:", X_reg.shape)
print("Categorical Features:", len(categorical_reg))
print("Numerical Features:", len(numerical_reg))


# Train-Test Split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)


print("\nRegression Training Data:", X_train_reg.shape)
print("Regression Testing Data:", X_test_reg.shape)
```

```python
# ============================================
# PREPROCESSING
# ============================================

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Classification preprocessing
classification_preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=True
            ),
            categorical_class
        ),
        (
            'numerical',
            'passthrough',
            numerical_class
        )
    ]
)


# Regression preprocessing
regression_preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=True
            ),
            categorical_reg
        ),
        (
            'numerical',
            'passthrough',
            numerical_reg
        )
    ]
)

print("Preprocessing setup completed successfully!")
```

```python
# ============================================
# STEP 6A: CLASSIFICATION MODELS
# ============================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Classification libraries imported successfully!")
```

```python
# ============================================
# LOGISTIC REGRESSION
# ============================================

logistic_model = Pipeline([
    ('preprocessor', classification_preprocessor),
    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])

print("Training Logistic Regression...")

logistic_model.fit(
    X_train_class,
    y_train_class
)

logistic_pred = logistic_model.predict(X_test_class)

print("Logistic Regression completed!")
```

```python
print("Logistic Regression Results")
print("=" * 40)

print("Accuracy:",
      accuracy_score(y_test_class, logistic_pred))

print("Precision:",
      precision_score(
          y_test_class,
          logistic_pred,
          average='weighted'
      ))

print("Recall:",
      recall_score(
          y_test_class,
          logistic_pred,
          average='weighted'
      ))

print("F1 Score:",
      f1_score(
          y_test_class,
          logistic_pred,
          average='weighted'
      ))

print("\nClassification Report:")
print(
    classification_report(
        y_test_class,
        logistic_pred
    )
)
```

```python
# ============================================
# RANDOM FOREST CLASSIFIER
# ============================================

rf_model = Pipeline([
    ('preprocessor', classification_preprocessor),
    ('model', RandomForestClassifier(
        n_estimators=100,
        max_depth=15,
        random_state=42,
        n_jobs=-1,
        class_weight='balanced'
    ))
])

print("Training Random Forest...")

rf_model.fit(
    X_train_class,
    y_train_class
)

rf_pred = rf_model.predict(X_test_class)

print("Random Forest completed!")
```

```python
print("Random Forest Results")
print("=" * 40)

print("Accuracy:",
      accuracy_score(y_test_class, rf_pred))

print("Precision:",
      precision_score(
          y_test_class,
          rf_pred,
          average='weighted'
      ))

print("Recall:",
      recall_score(
          y_test_class,
          rf_pred,
          average='weighted'
      ))

print("F1 Score:",
      f1_score(
          y_test_class,
          rf_pred,
          average='weighted'
      ))

print("\nClassification Report:")
print(
    classification_report(
        y_test_class,
        rf_pred
    )
)
```
